## adverse_media_search

```python
# Example setup
scores = [1.45, -0.2, 3.1, 0.0, -1.5, 2.8]
norm_scores = [item / max(scores) for item in scores]
k = 3

# 1. Keep (index, score) pairs only if the score is >= 0
filtered_pairs = [(idx, score) for idx, score in enumerate(scores) if score >= 0]

# 2. Sort the pairs based on the score in descending order (highest first)
sorted_pairs = sorted(filtered_pairs, key=lambda item: item[1], reverse=True)

# 3. Extract just the original indices for the top K items
final_indices = [idx for idx, score in sorted_pairs[:k]]
```

## ollama

```python
import requests
from typing import Dict, Any
```

```python
base_url = 'http://localhost:11434/'
model = 'qwen2.5:7b'
```

```python
base_url = base_url.rstrip("/")
model = model.strip()
```

```python
system = 'you are a helpful assistant to respond to user query below: {user}'

user = 'what is the capital of America and how far is it from capital of Canada in miles?'

max_tokens = 1024
temperature = 0

payload = {
    "model": model,
    "messages": [
        {"role": "system", "content": system},
        {"role": "user", "content": user}
    ],
    "max_tokens": max_tokens,
    "temperature": temperature,
    "stream": False
}

url = f"{base_url}/api/chat"
```

```python
response = requests.post(url, json=payload)
```

```python
data = response.json()
```
